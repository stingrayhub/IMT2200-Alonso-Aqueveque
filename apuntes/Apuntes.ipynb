# Merge:
La función `merge` permite combinar dos DataFrames en base a una o más columnas clave. Es útil para unir información de diferentes fuentes.

**Sintaxis básica:**
```python
## Una Columna Clave
pd.merge(df1, df2, on='columna_clave', how='inner')

## Dos o Más
pd.merge(df1, df2, on=['columna1', 'columna2'], how='inner')
```


- `on`: columna(s) por las que se realiza la unión.
- `how`: tipo de unión (`inner`, `outer`, `left`, `right`).
### **Ejemplo:**

Supongamos que tienes dos DataFrames con información de estudiantes y sus notas:
```python
import pandas as pd

**DataFrame estudiantes:**

| id | nombre |
|----|--------|
| 1  | Ana    |
| 2  | Luis   |
| 3  | Sofía  |

**DataFrame notas:**

| id | nota |
|----|------|
| 1  | 6.5  |
| 2  | 5.8  |
| 4  | 7.0  |

```

### Inner
```python
resultado = pd.merge(estudiantes, notas, on='id', how='inner')

**Resultado del merge:**

| id | nombre | nota |
|----|--------|------|
| 1  | Ana    | 6.5  |
| 2  | Luis   | 5.8  |

```

### Outer
```python
resultado_outer = pd.merge(estudiantes, notas, on='id', how='outer')

**Resultado del merge (outer):**

| id | nombre | nota |
|----|--------|------|
| 1  | Ana    | 6.5  |
| 2  | Luis   | 5.8  |
| 3  | Sofía  | NaN  |

```

### Left
```python

| id | nombre | nota |
|----|--------|------|
```


# Concat
La función `concat` permite unir DataFrames o Series de manera vertical (por filas) u horizontal (por columnas). Es útil para apilar datos o agregar nuevas variables.

**Sintaxis básica:**
```python
pd.concat([df1, df2], axis=0)  # Por filas
pd.concat([df1, df2], axis=1)  # Por columnas
# El default es por filas (`axis=0`)
```
- `axis=0`: concatena por filas.
- `axis=1`: concatena por columnas.


Si no se especifica el parámetro `axis`, la función `concat` une los DataFrames por filas. Es decir, agrega los registros uno debajo del otro.

Por ejemplo:
```python
pd.concat([df1, df2])
# Equivalente a pd.concat([df1, df2], axis=0)
```

**Ejemplo: Concatencación por filas**

Supón que tienes dos DataFrames con información de diferentes años:
```python

**DataFrame df_2024:**

| id | nota |
|----|------|
| 3  | 7.0  |
| 4  | 6.2  |

concat_df = pd.concat([df_2023, df_2024], axis=0)

**Resultado del concat:**

| id | nota |
|----|------|
| 1  | 6.5  |
| 2  | 5.8  |
| 3  | 7.0  |
| 4  | 6.2  |
```

**Ejemplo: Concatenación por columnas**

Supón que tienes dos DataFrames con información distinta para los mismos estudiantes:

```python
**DataFrame notas:**
```

## 3. GroupBy

La función `groupby` permite agrupar los datos según una o más columnas y aplicar funciones de agregación (como suma, promedio, conteo, etc.).

**Sintaxis básica:**
```python
df.groupby('columna').agg({'otra_columna': 'funcion'})
```
- `'columna'`: columna por la que se agrupa.
- `'funcion'`: función de agregación (`sum`, `mean`, `count`, `max`, `min`, `median`, etc.).

**Ejemplo:**
Supón que tienes un DataFrame con información de estudiantes y sus notas:
```python
**DataFrame datos:**

|-------|------|
| A     | 6.5  |
| A     | 5.8  |
| B     | 7.0  |
| B     | 6.2  |

**Resultado del groupby:**  

| curso | nota promedio |
|-------|--------------|
| A     | 6.15         |
| B     | 6.60         |

**Ejemplo: Suma de notas por curso**

suma_notas = datos.groupby('curso').agg({'nota': 'sum'})

**Resultado del groupby (suma):**

| curso | suma de notas |
|-------|--------------|
| A     | 12.3         |
| B     | 13.2         |

```
**Ejemplo: GroupBy con múltiples columnas**

Puedes agrupar por más de una columna para obtener estadísticas por combinación de valores:


## Extra: ``explode``

La función `explode` en pandas permite transformar una columna que contiene listas o arreglos en varias filas, expandiendo cada elemento de la lista en una fila separada. Es útil cuando tienes datos anidados y quieres "desempaquetarlos" para analizarlos más fácilmente.

**Sintaxis básica:**
```python
df.explode('columna')
```
- `'columna'`: nombre de la columna que contiene listas o arreglos.

**Ejemplo:**

Supón que tienes un DataFrame donde cada estudiante puede tener varias materias inscritas:

```python
import pandas as pd

datos = pd.DataFrame({
    'id': [1, 2, 3],
    'nombre': ['Ana', 'Luis', 'Sofía'],
    'materias': [['Matemáticas', 'Historia'], ['Física'], ['Química', 'Arte', 'Música']]
})

datos
```
**Resultado original:**
| id | nombre | materias                      |
|----|--------|------------------------------|
| 1  | Ana    | ['Matemáticas', 'Historia']  |
| 2  | Luis   | ['Física']                   |
| 3  | Sofía  | ['Química', 'Arte', 'Música']|

Ahora aplicamos `explode` sobre la columna `materias`:

```python
datos_exploded = datos.explode('materias')
datos_exploded
```

**Resultado después de explode:**
| id | nombre | materias    |
|----|--------|-------------|
| 1  | Ana    | Matemáticas |
| 1  | Ana    | Historia    |

| 3  | Sofía  | Química     |
| 3  | Sofía  | Arte        |

Cada materia ahora aparece en una fila separada, facilitando el análisis por materia.

## Algunas formas de ocupar Groupby

````python
# 1. Agrupación por una Sola Columna 📊
# Esta es la forma más común, donde agrupas todas las filas que tienen el mismo valor en una columna específica.


# Agrupa las filas que tienen el mismo valor en la columna 'tipo'
agrupado = df.groupby('tipo')

# Ejemplo de uso: calcular el promedio de 'ataque' para cada 'tipo'
promedios = df.groupby('tipo')['ataque'].mean()

# 2. Agrupación por Múltiples Columnas 🧩
# Puedes agrupar por una combinación de valores en varias columnas, creando grupos más específicos.


# Agrupa por la combinación única de valores en 'region' y 'tipo'
agrupado = df.groupby(['region', 'tipo'])

# Ejemplo de uso: calcular la suma de 'ventas' por cada combinación de 'region' y 'tipo'
sumas = df.groupby(['region', 'tipo'])['ventas'].sum()

# 3. Agrupación por Función o Diccionario ⚙️
# Puedes agrupar las filas basándose en la aplicación de una función al índice, o usando un diccionario que mapee los valores del índice a los nuevos grupos. Esto es muy útil cuando el índice contiene información útil (como fechas).

# Por Nivel de Índice (Si tienes un MultiIndex)
# Si tu DataFrame tiene un índice jerárquico (MultiIndex):


# Agrupa por los valores del primer nivel del índice (nivel 0)
agrupado_nivel = df.groupby(level=0)
# Por Función en el Índice (Común para fechas)

# Suponiendo que el índice es de tipo datetime:
# Agrupa por el año del índice (por ejemplo, para datos anuales)
df.groupby(df.index.year)['valor'].sum()

# 4. Aplicación de Múltiples Funciones (Aggregación) 🔢
# Una vez que has creado el objeto GroupBy, puedes aplicar diferentes funciones a las columnas utilizando el método .agg() o .aggregate().

# A. Múltiples Funciones en una Columna
# Aplica varias funciones estadísticas (suma, media, conteo) a una sola columna.


# Calcula el promedio, la suma y el conteo de la columna 'precio' por cada 'categoria'
resultado = df.groupby('categoria')['precio'].agg(['mean', 'sum', 'count'])
# B. Múltiples Funciones en Múltiples Columnas
# Aplica diferentes funciones a diferentes columnas en la misma agrupación.


# Agrupa por 'ciudad'
# A la columna 'temperatura' le aplica la media y el máx.
# A la columna 'humedad' le aplica el mínimo y el conteo.
resultado = df.groupby('ciudad').agg(
    promedio_temp=('temperatura', 'mean'),
    max_temp=('temperatura', 'max'),
    min_hum=('humedad', 'min'),
    conteo_hum=('humedad', 'count')
)

# 5. Aplicación Fila por Fila (Apply) 🧮
# El método .apply() es el más flexible, ya que te permite pasar una función personalizada (lambda o normal) a cada grupo.

# Agrupa por 'grupo' y aplica una función personalizada a cada grupo (sub-DataFrame)
def rango_valores(grupo):
    return grupo['valor_max'].max() - grupo['valor_min'].min()

resultados_personalizados = df.groupby('grupo').apply(rango_valores)
````
